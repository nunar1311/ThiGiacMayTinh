```python
import cv2
import time
```

```python
camera = cv2.VideoCapture(0)
```

```python
cv2.namedWindow("Played", cv2.WINDOW_NORMAL)

interval = 20
count = 0
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX
font_color = (0, 255, 0)
font_scale = 1
font_thicknes = 2
```

```python
while True:
    start_time = time.time()

    ret, frame = camera.read()
    if not ret:
        break

    count = count + 1

    if (count%interval) == 0:
        cv2.imwrite(f"../ThiGiacMayTinh/data/image_{count}.jpg", frame)

    end_time = time.time()

    fps = 1 / (end_time - start_time)

    cv2.putText(frame, f"FPS: {fps:.2f}", (10, 30), font, font_scale, font_color, font_thicknes)
    cv2.imshow("Video Played", frame)

    if cv2.waitKey(1) == ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```
