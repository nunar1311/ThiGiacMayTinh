```python
import cv2
```

```python
def display_image(image):
    cv2.imshow('image', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
face_cascade = cv2.CascadeClassifier('../ThiGiacMayTinh/data/haarcascade_frontalcatface.xml')
eye_cascade = cv2.CascadeClassifier('../ThiGiacMayTinh/data//haarcascade_eye_tree_eyeglasses.xml')
```

```python
img = cv2.imread('data/image_620.jpg')
```

```python
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
face = face_cascade.detectMultiScale(gray_image, scaleFactor=1.1, minNeighbors=5, minSize=(100, 100))
```

```python
eyes = eye_cascade.detectMultiScale(gray_image, scaleFactor=1.1, minNeighbors=1, minSize=(5,5))
```

```python
for(x, y, w, h) in face:
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)
```

```python
for(x, y, w, h) in eyes:
    cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2)
```

```python
display_image(img)
```
