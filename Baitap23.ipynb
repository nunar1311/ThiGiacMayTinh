```python
import cv2
import time
```

```python
camera = cv2.VideoCapture(0)
```

```python
cv2.namedWindow("Video played", cv2.WINDOW_NORMAL)
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX
font_color = (0, 255, 0)
font_scale = 1
font_thickness = 2
```

```python
while True:
    start_time = time.time()
    ref, frame = camera.read()

    if not ref:
        break

    end_time = time.time()

    fps = 1 / (end_time - start_time)

    cv2.putText(frame, f"FPS: {fps:.2f}", (10, 30), font, font_scale, font_color, font_thickness)
    cv2.imshow("Video played", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break


camera.release()
cv2.destroyAllWindows()
```
