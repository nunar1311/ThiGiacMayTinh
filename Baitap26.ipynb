```python
import cv2
import time
```

```python
def detect(img):
    face_cascade = cv2.CascadeClassifier("../ThiGiacMayTinh/data/haarcascade_frontalcatface.xml")
    eye_cascade = cv2.CascadeClassifier("../ThiGiacMayTinh/data/haarcascade_eye_tree_eyeglasses.xml")

    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray_img, scaleFactor=1.1, minNeighbors=5, minSize=(30,30))
    eyes = eye_cascade.detectMultiScale(gray_img, scaleFactor=1.1, minNeighbors=1, minSize=(5,5))

    for (x, y, w, h) in faces:
        cv2.rectangle(img, (x,y), (x+w, y+h), (255,0,0), 2)

    for (x, y, w, h) in eyes:
        cv2.rectangle(img, (x,y), (x+w, y+h), (0,255,0), 2)

    return img
```

```python
camera = cv2.VideoCapture(0)

cv2.namedWindow("Camera", cv2.WINDOW_NORMAL)

font = cv2.FONT_HERSHEY_SIMPLEX
font_color = (0, 255, 0)
font_scale = 1
font_thickness = 2

while True:
    start_time = time.time()
    ret, frame = camera.read()
    if not ret:
        break
    end_time = time.time()

    fps = 1/(end_time - start_time)
    frame = detect(frame)
    cv2.putText(frame, f'FPS: {int(fps)}', (10, 30), font, font_scale, font_color, font_thickness)
    cv2.imshow("Camera", frame)
    if cv2.waitKey(1) == ord('q'):
        break

camera.release()
cv2.destroyAllWindows()
```
